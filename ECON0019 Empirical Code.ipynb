```python
import pandas as pd
import statsmodels.formula.api as smf
import numpy as np
import linearmodels as plm
```

```python
df = pd.read_excel('ECON0019_empirical_project_2026 (1).xlsx')
```

```python
df = df.set_index (['state', 'year'], drop=False)
```

```python
# 2(I) Code
reg_2I = smf.ols(formula='np.log(viorate) ~ 1 + shallcarry', data=df)
results_2I = reg_2I.fit(cov_type="cluster", cov_kwds={"groups": df["state"]})
print(f'2(I) Summary results: \n{results_2I.summary()}\n')
```

```python
# 2(II) Code
reg_2II = smf.ols(formula='np.log(viorate) ~ shallcarry + popdens + avginc + popul + pctmale + pctwhite + pctblack', data=df)
results_2II = reg_2II.fit(cov_type="cluster", cov_kwds={"groups": df["state"]})
print(f'2(II) Summary Results: \n{results_2II.summary()}\n')
```

```python
# Q3 Code
df_q3 = df.reset_index(drop=True)
df_q3['shallcarry_lag1'] = df_q3.groupby('state')['shallcarry'].shift(1)
df_q3 = df_q3.dropna(subset=['shallcarry_lag1'])

reg_3 = smf.ols(formula='np.log(viorate) ~ shallcarry + shallcarry_lag1 + popdens + avginc + popul + pctmale + pctwhite + pctblack', data=df_q3)
results_3 = reg_3.fit(cov_type='cluster', cov_kwds={'groups': df_q3['state']})
print(f'Q3 Summary Results: \n{results_3.summary()}\n')
```

```python
# Q4 Code
df = df.set_index(['state', 'year'], drop=False)

# Fixed Effects Estimator
reg_4 = plm.PanelOLS.from_formula('np.log(viorate) ~ shallcarry + popdens + avginc + popul + pctmale + pctwhite + pctblack + EntityEffects', data=df, drop_absorbed=True)
results_4 = reg_4.fit(cov_type='clustered', cluster_entity=True)
print(f'Q4 FE Summary Results: \n{results_4.summary}\n')
```

```python
# Q5 Code
df= df.set_index(['state', 'year'], drop=False)

# Fixed and Time Effects Estimator
reg_5 = plm.PanelOLS.from_formula('np.log(viorate) ~ shallcarry + popdens + avginc + popul + pctmale + pctwhite + pctblack + EntityEffects + TimeEffects', data=df, drop_absorbed=True)
results_5 = reg_5.fit(cov_type='clustered', cluster_entity=True)
print(f'Q5 Fixed and Time Effects Summary Results: \n{results_5.summary}\n')
```
