```python
import pandas as pd
import statsmodels.formula.api as smf
import numpy as np
```

```python
df = pd.read_excel('ECON0019_empirical_project_2026 (1).xlsx')
```

```python
# Generating logviorate variable
df['logviorate'] = np.log(df['viorate'])
```

```python
# 2(I) Code
print(f'2(I) Summary results: \n{smf.ols(formula="logviorate ~ shallcarry", data=df).fit().summary()}\n')
```

```python
# 2(II) Code
print(f'2(II) Summary Results: \n{smf.ols(formula="logviorate ~ shallcarry + popdens + avginc + popul + pctmale + pctwhite + pctblack", data=df).fit().summary()}\n')
```
