```python
import pandas as pd
df = pd.read_excel('ECON0019_empirical_project_2026 (1).xlsx')
```

```python
import numpy as np
df['logviorate'] = np.log(df['viorate'])
```

```python
import statsmodels.formula.api as smf

# 2(I) Code
reg = smf.ols(formula='logviorate ~ shallcarry', data=df).fit()
print(f'2(I) Summary Results: \n{reg.summary()}\n')
print(f'2(I) Parameters: \n{reg.params}\n')

```

```python
# 2(II) Code
reg = smf.ols(formula='logviorate ~ shallcarry + popdens + avginc + popul + pctmale + pctwhite + pctblack', data=df).fit()
print(f'2(II) Summary Results: \n{reg.summary()}\n')
print(f'2(II) Parameters: \n{reg.params}\n')
```
