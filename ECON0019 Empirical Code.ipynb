```python
import pandas as pd
import statsmodels.formula.api as smf
import numpy as np
import linearmodels as plm
```

```python
df = pd.read_excel('ECON0019_empirical_project_2026 (1).xlsx')
```

```python
df = df.set_index (['state', 'year'], drop=False)
```

```python
# 2(I) Code
reg = smf.ols(formula='np.log(viorate) ~ 1 + shallcarry', data=df)
results = reg.fit(cov_type="cluster", cov_kwds={"groups": df["state"]})
print(f'2(I) Summary results: \n{results.summary()}\n')
```

```python
# 2(II) Code
reg = smf.ols(formula='np.log(viorate) ~ shallcarry + popdens + avginc + popul + pctmale + pctwhite + pctblack', data=df)
results = reg.fit(cov_type="cluster", cov_kwds={"groups": df["state"]})
print(f'2(II) Summary Results: \n{results.summary()}\n')
```

```python
# Q3 Code
df['shallcarry_lag1'] = df['shallcarry'].shift(1)

reg = smf.ols(formula='np.log(viorate) ~ shallcarry + shallcarry_lag1 + popdens + avginc + popul + pctmale + pctwhite + pctblack', data=df)
results = reg.fit(cov_type="cluster", cov_kwds={"groups": df["state"]})
print(f'Q3 Summary Results: \n{results.summary()}\n')
```
